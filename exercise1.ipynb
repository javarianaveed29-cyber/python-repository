```python
def string_operations(s):
    reversed_str = s[::-1]
    every_second_char = s[1::2]
    return reversed_str, every_second_char

# Example
print(string_operations("hello"))
```

```python
def even_squares(lst):
    return [x**2 for x in lst if x % 2 == 0]

print(even_squares([1,2,3,4,5,6]))
```

```python
def char_frequency(s):
    freq = {}
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
    return freq

print(char_frequency("hello"))
```

```python
def odd_indices(lst):
    return [i for i, val in enumerate(lst) if val % 2 != 0]

print(odd_indices([10,11,12,13,14]))
```

```python
def rotate_right(lst, k):
    if len(lst) == 0:
        return lst
    k = k % len(lst)
    return lst[-k:] + lst[:-k]

print(rotate_right([1,2,3,4,5], 2))
```

```python
def list_to_dict(keys, values):
    return dict(zip(keys, values))

print(list_to_dict(["a","b","c"], [1,2,3]))
```

```python
def manual_iteration(lst):
    it = iter(lst)
    
    while True:
        try:
            print(next(it))
        except StopIteration:
            print("Iteration completed")
            break

manual_iteration([1,2,3])
```

```python
def flatten(nested):
    return [item for sublist in nested for item in sublist]

print(flatten([[1,2],[3,4],[5,6]]))
```

```python
def is_palindrome(s):
    return s == s[::-1]

print(is_palindrome("madam"))
print(is_palindrome("hello"))
```

```python
def second_largest(lst):
    if len(lst) < 2:
        return None

    first = second = float('-inf')

    for num in lst:
        if num > first:
            second = first
            first = num
        elif first > num > second:
            second = num

    return second

print(second_largest([10, 20, 4, 45, 99]))
```
